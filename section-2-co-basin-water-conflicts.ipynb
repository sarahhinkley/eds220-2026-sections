```python
# One state per row (with extra spaces to be corrected)
states_notyetunique = (df.loc[df['Conflict Present']=='Y', "State"]
    .str.split(';') # Break string at ;
    .explode() # Expand lists for one state per row
    .unique()
)
states_notyetunique
```

```python
states_notyetunique.shape
```

```python
# One state per row with no extra spaces
states_unique = (df.loc[df['Conflict Present']=='Y', "State"]
    .str.split(';') # Break string at ;
    .explode() # Expand lists for one state per row
    .str.strip() # Remove leading/trailing white space
    .unique()
)
states_unique
```

```python
states_unique.shape
```

```python
# One state per row with NA values kept to make NaN a column in bar plot
states_unique_keepNA = (df.loc[df['Conflict Present']=='Y', "State"]
    .str.split(';') # Break string at ;
    .explode() # Expand lists for one state per row
    .str.strip() # Remove leading/trailing white space
    .value_counts(dropna=False) # Keeps the NA values
)

states_unique_keepNA.plot(
    kind='bar',
    xlabel = 'Number of conflicts', 
    ylabel = 'State',
    title = 'Water conflicts in the Colorado River Basin by State'
)
```
